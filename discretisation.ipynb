<h1>APPLICATION DES MÉTHODES DE DISCRÉTISATION SIR LES DONNÉES DU DATASET UCI WINE</h1>

<h3>Importation des bibliothèques nécessaires</h3>

```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import KBinsDiscretizer
```

<h3>Chargement des données du dataset</h3>

```python
colonnes = [
    "Class",
    "Alcohol",
    "Malic_acid",
    "Ash",
    "Alcalinity_of_ash",
    "Magnesium",
    "Total_phenols",
    "Flavanoids",
    "Nonflavanoid_phenols",
    "Proanthocyanins",
    "Color_intensity",
    "Hue",
    "OD280/OD315_of_diluted_wines",
    "Proline"
]

df = pd.read_csv("wine.data", header=None, names=colonnes)

# Variables numériques à discrétiser (toutes sauf Class)
features = df.drop(columns=["Class"])
```

<h2>MÉTHODE DE DISCRÉTISATION A LARGEUR ÉGALE</h2>

```python
# Discrétisation à largeur égale avec affichage des intervalles
for col in features.columns:
    min_val = features[col].min()
    max_val = features[col].max()
    width = (max_val - min_val) / 3
    
    # Construction des intervalles
    intervals = [
        (min_val, min_val + width),
        (min_val + width, min_val + 2*width),
        (min_val + 2*width, max_val)
    ]
    
    # Discrétisation avec pd.cut
    df[col + "_bin"] = pd.cut(features[col], bins=[min_val, min_val + width, min_val + 2*width, max_val], labels=[0, 1, 2], include_lowest=True)
    
    print(f"\nVariable : {col}")
    print(f"Min : {min_val:.3f}, Max : {max_val:.3f}, Largeur intervalle : {width:.3f}")
    print("Intervalles obtenus :")
    for i, interval in enumerate(intervals):
        print(f"  Intervalle {i+1}: [{interval[0]:.3f} ; {interval[1]:.3f}[")

# Affichage des 5 premières lignes du dataframe avec les colonnes discrétisées
print("\nExemple de discrétisation à largeur égale :")
print(df[[col+"_bin" for col in features.columns]].head())
```

<h2>MÉTHODE DE DISCRÉTISATION A FREQUENCE ÉGALE</h2>

```python

# Discrétisation à fréquence égale (3 intervalles)
for col in features.columns:
    sorted_values = np.sort(features[col].values)
    N = len(sorted_values)
    N_interval = N // 3  # nombre d'observations par intervalle
    
    # Détermination des bornes
    bins = [
        sorted_values[0], 
        sorted_values[N_interval - 1], 
        sorted_values[2*N_interval - 1], 
        sorted_values[-1]
    ]
    
    # Pour s'assurer que toutes les valeurs sont incluses
    bins[1] += 1e-8
    bins[2] += 1e-8
    
    # Discrétisation avec pd.cut
    df[col + "_qbin"] = pd.cut(features[col], bins=bins, labels=[0,1,2], include_lowest=True)
    
    print(f"\nVariable : {col}")
    print("Nombre d'observations :", N)
    print("Observations par intervalle :", N_interval)
    print("Intervalles obtenus :")
    for i in range(3):
        print(f"  Intervalle {i+1}: [{bins[i]:.3f} ; {bins[i+1]:.3f}[")

# Affichage des 5 premières lignes des colonnes discrétisées
print("\nExemple de discrétisation à fréquence égale :")
print(df[[col+"_qbin" for col in features.columns]].head())
```

<h2>MÉTHODE DE DISCRÉTISATION AVEC L'ENTROPIE ( Sans utilisation de fonction prédéfinis)</h2>

```python


# Fonction pour calculer l'entropie d'une série de classes
def entropy(y):
    probs = y.value_counts(normalize=True)
    return -np.sum(probs * np.log2(probs + 1e-9))  # ajouter petit epsilon pour éviter log(0)


features = df.drop(columns=["Class"])
y = df["Class"]

# Discrétisation par entropie (2 intervalles)
entropy_disc = pd.DataFrame()

for col in features.columns:
    sorted_values = np.sort(features[col].values)
    best_entropy = float('inf')
    best_threshold = None
    
    for i in range(1, len(sorted_values)):
        threshold = sorted_values[i]
        left_classes = y[features[col] <= threshold]
        right_classes = y[features[col] > threshold]
        
        H_left = entropy(left_classes)
        H_right = entropy(right_classes)
        
        p_left = len(left_classes) / len(y)
        p_right = len(right_classes) / len(y)
        
        H_total = H_left * p_left + H_right * p_right
        
        if H_total < best_entropy:  # on cherche le seuil qui minimise l'entropie
            best_entropy = H_total
            best_threshold = threshold
    
    # Création des intervalles
    bins = [-float('inf'), best_threshold, float('inf')]
    entropy_disc[col + "_ebin"] = pd.cut(features[col], bins=bins, labels=[0,1])
    
    print(f"\nVariable : {col}")
    print(f"Seuil retenu par entropie : {best_threshold:.3f}")
    print(f"Intervalles obtenus : [-inf, {best_threshold:.3f}], [{best_threshold:.3f}, +inf]")

# Affichage des 5 premières lignes des colonnes discrétisées
print("\nExemple de discrétisation par entropie :")
print(entropy_disc.head())
```

<h2>MÉTHODE DE DISCRÉTISATION AVEC L'ENTROPIE ( En utilisant l'arbre de décision)</h2>

```python

features = df.drop(columns=["Class"])
y = df['Class']  # variable cible

# --------------------------
# Discrétisation par entropie (2 intervalles)
# --------------------------
entropy_disc = pd.DataFrame()

for col in features.columns:
    # Arbre de décision sur une seule variable
    dt = DecisionTreeClassifier(criterion='entropy', max_leaf_nodes=2)
    X_col = features[[col]]
    dt.fit(X_col, y)
    
    # Seuils retenus par l'arbre
    thresholds = dt.tree_.threshold
    thresholds = thresholds[thresholds != -2]  # -2 correspond aux feuilles
    
    # Création des bins
    bins = [-float('inf')] + list(thresholds) + [float('inf')]
    entropy_disc[col + "_ebin"] = pd.cut(features[col], bins=bins, labels=[0,1])
    
    # Affichage des intervalles
    print(f"\nVariable : {col}")
    print(f"Seuil retenu par entropie : {thresholds[0]:.3f}")
    print(f"Intervalles obtenus : [-inf, {thresholds[0]:.3f}], [{thresholds[0]:.3f}, +inf]")

# Affichage des 5 premières lignes des colonnes discrétisées
print("\nExemple de discrétisation par entropie :")
print(entropy_disc.head())
```
