<h1>ANALYSE EXPLORATOIRE DES DONNÉES DU DATASET UCI WINE DE KAGGLE</h1>

<h3>Importation des bibliothèques</h3>

```python
import pandas as pd
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

<h3>Chargement du dataset</h3>

```python
# chemin du fichier téléchargé depuis Kaggle
fichier = "wine.data"  # ou "wine.csv"

# colonnes définies par l'UCI Wine Dataset
colonnes = [
    "Class",
    "Alcohol",
    "Malic_acid",
    "Ash",
    "Alcalinity_of_ash",
    "Magnesium",
    "Total_phenols",
    "Flavanoids",
    "Nonflavanoid_phenols",
    "Proanthocyanins",
    "Color_intensity",
    "Hue",
    "OD280/OD315_of_diluted_wines",
    "Proline"
]

# lecture du fichier
df = pd.read_csv(fichier, header=None, names=colonnes)
```

<h3>Affichage des 5 premières observations du dataset</h3>

```python
print(df.head())
```

<h3>Mesures statistiques des données</h3>

```python
print(df.describe())
```

<h3>Boîtes à moustache de chaque variable du dataset</h3>

```python
# On enlève la variable cible (Class)
features = df.drop(columns=["Class"])

# Création d'un boxplot pour toutes les variables
plt.figure(figsize=(14, 8))
sns.boxplot(data=features, orient="v")  # vertical => variables sur x, valeurs sur y
plt.xticks(rotation=45)  # rotation pour lisibilité des noms
plt.title("Boîtes à moustaches des variables du dataset Wine")
plt.tight_layout()
plt.show()
```

<h3>Matrice de corrélation entre les variables</h3>

```python
# On enlève la variable cible (Class)
features = df.drop(columns=["Class"])

# Matrice de corrélation
corr = features.corr()

# Afficher la matrice numérique
#print("Matrice de corrélation :\n", corr)

# Heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", cbar=True)
plt.title("Matrice de corrélation des variables du dataset Wine")
plt.tight_layout()
plt.show()
```

<h3>Histogramme de chaque variable</h3>

```python
features.hist(figsize=(16, 12), bins=15)
plt.suptitle("Histogrammes des variables")
plt.show()
```
